```python
import torch
import pandas as pd
import numpy as np
from PIL import Image
from vietocr.tool.predictor import Predictor
from vietocr.tool.config import Cfg
from faster_whisper import WhisperModel, BatchedInferencePipeline

print("="*60)
print("🧪 ĐANG TEST CHẠY THỰC TẾ TRÊN GPU...")
print("="*60)

# 1. Test VietOCR trên GPU 0
try:
    print("1️⃣ Test VietOCR GPU 0...")
    config = Cfg.load_config_from_name('vgg_transformer')
    config['device'] = 'cuda:0'
    config['predictor']['beamsearch'] = False
    detector = Predictor(config)
    dummy_img = Image.new('RGB', (100, 32), color=(255, 255, 255))
    _ = detector.predict(dummy_img)
    print("   👉 ✅ VietOCR + CRAFT chạy THỰC TẾ thành công trên GPU 0!")
except Exception as e:
    print(f"   👉 ❌ Lỗi VietOCR: {e}")

# 2. Test PhoWhisper trên GPU 1 (hoặc GPU 0)
try:
    dev_idx = 1 if torch.cuda.device_count() > 1 else 0
    print(f"2️⃣ Test PhoWhisper trên GPU {dev_idx}...")
    model = WhisperModel("kiendt/PhoWhisper-large-ct2", device="cuda", device_index=dev_idx, compute_type="float16")
    pipeline = BatchedInferencePipeline(model=model)
    print(f"   👉 ✅ PhoWhisper Large Batch Pipeline chạy THỰC TẾ thành công trên GPU {dev_idx}!")
except Exception as e:
    print(f"   👉 ❌ Lỗi PhoWhisper: {e}")

# 3. Test ghi file Parquet
try:
    test_df = pd.DataFrame([{"test": 1, "status": "ok"}])
    test_df.to_parquet("/kaggle/working/test_write.parquet")
    print("3️⃣ 👉 ✅ Quyền ghi đĩa Parquet Output: HOÀN HẢO!")
except Exception as e:
    print(f"3️⃣ 👉 ❌ Lỗi ghi file: {e}")

print("="*60)
print("🎉 TẤT CẢ ĐỀU XANH LÈ! SẴN SÀNG ĐỂ CHẠY CHÍNH THỨC!")
print("="*60)
```

```python
import subprocess
import os
import sys

output_ocr = "/kaggle/working/AIC2026/data/batch_1/processed/ocr_results.parquet"
output_asr = "/kaggle/working/AIC2026/data/batch_1/processed/transcripts.parquet"

# GPU 0: Chạy VietOCR (Bắt đầu tiếp từ Gói 5)
env_ocr = os.environ.copy()
env_ocr["CUDA_VISIBLE_DEVICES"] = "0"
cmd_ocr = [
    sys.executable, "scripts/extract_ocr_from_drive.py",
    "--urls_file", "config/drive_keyframes_urls.txt",
    "--output_path", output_ocr,
    "--crop_batch_size", "64",
    "--frame_batch_size", "16",
    "--use_vietocr",
    "--use_gpu",
    "--start_index", "5"
]

# GPU 1: Chạy PhoWhisper ASR (Bắt đầu tiếp từ Gói 6)
env_asr = os.environ.copy()
env_asr["CUDA_VISIBLE_DEVICES"] = "1"
cmd_asr = [
    sys.executable, "scripts/extract_asr_from_drive.py",
    "--urls_file", "config/drive_videos_urls.txt",
    "--output_path", output_asr,
    "--model_size", "vinai/PhoWhisper-large",
    "--batch_size", "16",
    "--beam_size", "5",
    "--compute_type", "float16",
    "--device", "cuda",
    "--start_index", "6"
]

p_ocr = subprocess.Popen(cmd_ocr, env=env_ocr)
p_asr = subprocess.Popen(cmd_asr, env=env_asr)

print("⚡ GPU 0 (VietOCR) và GPU 1 (PhoWhisper ASR) ĐANG CHẠY SONG SONG...")
p_ocr.wait()
p_asr.wait()

!cp /kaggle/working/AIC2026/data/batch_1/processed/*.parquet /kaggle/working/ 2>/dev/null || true
print("🎉🎉🎉 HOÀN THÀNH 100% TOÀN BỘ DỮ LIỆU!")
```
